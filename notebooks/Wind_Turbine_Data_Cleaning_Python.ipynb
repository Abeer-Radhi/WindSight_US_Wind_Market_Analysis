```python
# Create a backup of the dataframe before applying cleaning changes
# so the original state can later be compared with the cleaned dataset
df_before_cleaning = df.copy()

# Record the dataset dimensions before cleaning
print("Rows before cleaning:", len(df))
print("Columns:", len(df.columns))
```

```python
# Define the text columns that require formatting consistency
text_cols = [
    "state",
    "county",
    "project_name",
    "turbine_manufacturer",
    "turbine_model"
]

# Remove leading and trailing whitespace from text values
# to prevent duplicate categories caused by accidental spaces
for col in text_cols:
    df[col] = df[col].str.strip()
```

```python
# Convert empty text values into standard missing values (NaN)
# so missing text data is represented consistently across the dataset
import numpy as np

df[text_cols] = df[text_cols].replace(
    r"^\s*$",
    np.nan,
    regex=True
)
```

```python
# Convert key numerical fields to numeric data types to ensure they can be
# correctly used in calculations, comparisons, and statistical analysis
numeric_cols = [
    "turbine_id",
    "plant_id",
    "project_year",
    "turbine_capacity_kw",
    "retrofit_year",
    "longitude",
    "latitude"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
# Identify turbines where the recorded retrofit year occurs before the project year
# to detect logically inconsistent retrofit dates
retrofit_before_project = df[
    df["retrofit_year"].notna()
    & df["project_year"].notna()
    & (df["retrofit_year"] < df["project_year"])
]

# Count the number of records with inconsistent retrofit dates
print(
    "Retrofit years earlier than project year:",
    len(retrofit_before_project)
)

# Display up to 20 inconsistent records for further investigation
retrofit_before_project[
    [
        "turbine_id",
        "project_name",
        "project_year",
        "is_retrofitted",
        "retrofit_year"
    ]
].head(20)
```

```python
# Validate latitude and longitude values to identify coordinates
# outside the valid geographic ranges before mapping the turbines
invalid_coordinates = df[
    ~df["latitude"].between(-90, 90)
    | ~df["longitude"].between(-180, 180)
]

print("Invalid coordinates:", len(invalid_coordinates))
```

```python
# Define the priority fields used in the project's main analysis
priority_cols = [
    "turbine_id",
    "plant_id",
    "state",
    "county",
    "project_name",
    "project_year",
    "turbine_manufacturer",
    "turbine_model",
    "turbine_capacity_kw",
    "is_retrofitted",
    "retrofit_year",
    "longitude",
    "latitude"
]

# Create a final data-quality summary for the priority fields
# showing data types, missing values, missing percentages, and unique values
quality_report = pd.DataFrame({
    "data_type": df[priority_cols].dtypes.astype(str),
    "missing_count": df[priority_cols].isna().sum(),
    "missing_percent": (
        df[priority_cols].isna().mean() * 100
    ).round(2),
    "unique_values": df[priority_cols].nunique()
})

quality_report
```

```python
# Import Pandas for  (DataFrames, filtering, aggregation, and Excel file reading)
import pandas as pd

# Import Numpy for fast numerical operations, mathematical arrays, and missing value management (NaNs)
import numpy as np

# File path to the dataset
file_path = "EIA923_Schedules_2_3_4_5_M_12_2020_Final_Revision.xlsx"

# Initial load: Reading the excel file normally to inspect the structure
# Noticed that top rows contain title metadata and notes instead of column headers
df_temp = pd.read_excel(file_path, sheet_name="Page 1 Generation and Fuel Data")
print(df_temp)

# Reload data: Skip top metadata rows and set correct header row (row index 5)
df_gen = pd.read_excel(
    file_path, 
    sheet_name="Page 1 Generation and Fuel Data", 
    header=5
)

# Print dataset dimensions (rows, columns) to confirm data loaded properly
print("Dataset shape after fixing header:", df_gen.shape)

# Display the first 5 rows to verify clean structure
df_gen.head()
```

```python
# Clean column headers by stripping whitespace and inline newlines
df_gen.columns = df_gen.columns.str.replace('\n', ' ').str.strip()

# Drop rows and columns that are entirely blank
df_cleaned = df_gen.dropna(how='all', axis=0).dropna(how='all', axis=1)

# Filter dataset for Wind generation records only ('WND')
df_wind = df_cleaned[df_cleaned['Reported Fuel Type Code'] == 'WND'].copy()

# Inspect filtered subset dimensions
print(f"Wind generation subset: {len(df_wind)} rows, {df_wind.shape[1]} columns")
df_wind.head()
```

```python
# Drop irrelevant column for wind
if 'Physical Unit Label' in df_wind.columns:
    df_wind.drop(columns=['Physical Unit Label'], inplace=True)

# Impute missing categorical attributes with 'Unknown'
cat_cols = ['NERC Region', 'Balancing Authority Code', 'Respondent Frequency']
for col in cat_cols:
    if col in df_wind.columns:
        df_wind[col] = df_wind[col].fillna('Unknown')

# Check remaining missing values across columns
missing_counts = df_wind.isnull().sum()
print("Columns with missing values:")
print(missing_counts[missing_counts > 0])
```

```python
# Identify monthly net generation and key numerical fields
monthly_cols = [col for col in df_wind.columns if col.startswith('Netgen ')]
target_numeric = monthly_cols + ['Net Generation (Megawatthours)', 'Total Fuel Consumption MMBtu', 'Elec_MMBtu Year To Date']

# Filter for numeric columns present in dataframe
cols_to_convert = [col for col in target_numeric if col in df_wind.columns]

# Convert monthly generation values to float and replace invalid/empty entries with 0.0
for col in cols_to_convert:
    df_wind[col] = pd.to_numeric(df_wind[col], errors='coerce').fillna(0.0)

# Format Plant Id as integer
if 'Plant Id' in df_wind.columns:
    df_wind['Plant Id'] = df_wind['Plant Id'].astype(int)

# Verify data types after conversion
print(df_wind[cols_to_convert[:3]].dtypes)
```

```python
# Remove exact duplicate rows across all fields
df_wind = df_wind.drop_duplicates().copy()

# Primary key attributes for grouping and state-level aggregation
group_keys = [
    'Plant Id', 
    'Plant Name', 
    'Operator Id', 
    'Operator Name', 
    'Plant State', 
    'NERC Region', 
    'Balancing Authority Code', 
    'Reported Fuel Type Code', 
    'Reported Prime Mover'
]

# Keep only existing group keys in dataframe
existing_keys = [col for col in group_keys if col in df_wind.columns]

# Group records by plant identifiers and sum numeric generation values
df_wind = df_wind.groupby(existing_keys, as_index=False)[cols_to_convert].sum()

print(f"Aggregated record count: {len(df_wind)} rows")
```

```python
# Select relevant analytical columns for final clean export
keep_cols = [
    'Plant Id', 
    'Plant Name', 
    'Operator Id',
    'Operator Name', 
    'Plant State', 
    'NERC Region', 
    'Balancing Authority Code', 
    'Reported Fuel Type Code', 
    'Reported Prime Mover',
    'Netgen January', 'Netgen February', 'Netgen March', 'Netgen April',
    'Netgen May', 'Netgen June', 'Netgen July', 'Netgen August',
    'Netgen September', 'Netgen October', 'Netgen November', 'Netgen December',
    'Net Generation (Megawatthours)'
]

# Retain existing target columns
df_wind = df_wind[[col for col in keep_cols if col in df_wind.columns]].copy()

# Preview clean dataset structure
print(f"Final shape: {df_wind.shape[0]} rows x {df_wind.shape[1]} columns")
df_wind.head(3)
```

```python
# Finally, time to export the cleaned dataset
df_wind.to_csv("EIA923_2020_Wind_Cleaned.csv", index=False)
df_wind.to_excel("EIA923_2020_Wind_Cleaned.xlsx", index=False)

print("Export completed successfully.")
```
